```python
import os
import json
import torch
from google.colab import drive
from langchain_core.documents import Document
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer, DataCollatorForLanguageModeling
from peft import LoraConfig, get_peft_model
from captum.attr import LayerIntegratedGradients
from datasets import Dataset
from datetime import datetime

# 경로 기본 세팅
project_path = "/content/drive/MyDrive/MyProject"
save_path = os.path.join(project_path, "saved_qwen3b")

# ==========================================
# 1단계: JSONL 데이터 로드 및 지식 창고 생성
# ==========================================
print("임베딩 모델을 준비합니다...")
embeddings = HuggingFaceEmbeddings(model_name="BAAI/bge-m3")                  #🚨 [수정]  기존  snunlp 모델->  BAAI/bge-m3 모델로 변경

# 데이터 경로 (이중 폴더 주의)
jsonl_file_path = os.path.join(project_path, "data", "welfare_chunks.jsonl")  # 🔄️ 기존 data1/chunks.jsonl에서 바꿨습니다.

docs = []
print("JSONL 파일에서 데이터를 읽어옵니다...")

try:
    with open(jsonl_file_path, "r", encoding="utf-8") as f:
        for line in f:
            item = json.loads(line)
            doc = Document(
                page_content=item["text"], # 🔄️ payload 교체했습니다.
                metadata={
                    "제목": item.get("policy_name", ""),
                    "설명 구분": item.get("section", ""),
                    "신청 링크": item.get("url", ""),
                    "필터링": item.get("region", "")
                }
            )
            docs.append(doc)
    print(f"완료: 총 {len(docs)}개의 문서 청크를 불러왔습니다.")

except Exception as e:
    print(f"데이터 로드 실패: {e}")

if docs:
    print("지식 창고(FAISS)를 구축 중입니다...")
    vectorstore = FAISS.from_documents(documents=docs, embedding=embeddings)
    faiss_save_path = os.path.join(project_path, "saved_faiss_index_json")
    vectorstore.save_local(faiss_save_path)
    print("지식 창고 저장 완료")

# 🚨 안전장치: 데이터가 없으면 중단
if len(docs) == 0:
    raise RuntimeError("데이터를 찾을 수 없습니다. 경로를 확인해주세요.")

# ==========================================
# 2단계: AI 모델 및 XAI(Captum) 세팅
# ==========================================
print("2단계: Qwen 모델 및 가중치 세팅 중...")
model_id = "Qwen/Qwen2.5-3B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_id)
base_model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.float16, device_map={"": 0})    # 🚨 [수정] 기존 device_map "auto"->  device_map {"": 0}으로 변경,, GPU 메모리 때문

# 🔄️ 다시 실행해도 파인튜닝하지 않아도 되도록 수정
needs_training = True

# 모델 구조가 다를 경우 기존 가중치를 불러오지 않도록 처리
if os.path.exists(os.path.join(save_path, "adapter_config.json")):
    print("기존 학습 가중치를 불러옵니다. (파인튜닝을 생략합니다)")
    from peft import PeftModel
    model = PeftModel.from_pretrained(base_model, save_path, is_trainable=False)
    needs_training = False # 가중치가 있으니 학습 불필요!
else:
    print("새로운 LoRA 어댑터를 생성합니다. (파인튜닝을 진행합니다)")
    # Qwen 모델에 최적화된 타겟 모듈 설정
    config = LoraConfig(
        r=16,
        lora_alpha=32,
        target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],
        task_type="CAUSAL_LM"
    )
    model = get_peft_model(base_model, config)

# ✅ Qwen 모델 구조에 맞게 Captum 타겟 레이어 경로 수정
target_layer = model.base_model.model.model.embed_tokens
lig = LayerIntegratedGradients(model, target_layer)

# ==========================================
# 🔄️ 2.5단계 & 2.6단계: 가중치가 없을 때만 학습 진행
# ==========================================
if needs_training:
    print("학습용 데이터 변환 중...")
    train_texts = [doc.page_content for doc in docs]
    dataset = Dataset.from_dict({"text": train_texts})

    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    def tokenize_function(examples):
        return tokenizer(examples["text"], truncation=True, padding="max_length", max_length=256)

    tokenized_datasets = dataset.map(tokenize_function, batched=True)
    data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

    # 체크포인트 경로 설정
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    checkpoint_dir = os.path.join(project_path, f"checkpoints_{timestamp}")

    training_args = TrainingArguments(
        output_dir=checkpoint_dir,
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        num_train_epochs=3,
        learning_rate=2e-4,
        fp16=True,
        save_strategy="epoch",
        logging_steps=10,
        save_total_limit=2,
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_datasets,
        data_collator=data_collator,
    )

    print("모델 파인튜닝 시작")
    trainer.train()

    # 최종 모델 저장
    if not os.path.exists(save_path):
        os.makedirs(save_path)
    trainer.model.save_pretrained(save_path)
    tokenizer.save_pretrained(save_path)
    print("학습 및 저장 완료")
else:
    print("이미 완성된 뇌(가중치)가 있으므로 2.5단계와 2.6단계(재학습)를 건너뜁니다!")
```

```python
import re
from qdrant_client import QdrantClient
from langchain_qdrant import QdrantVectorStore
from langchain_community.embeddings import HuggingFaceEmbeddings
from google.colab import userdata   # 🔄️colab 보안에서 api key와 token 불러오기

# =====================================================================
# [단계 1~4] Qdrant DB 연결
# =====================================================================
embeddings = HuggingFaceEmbeddings(
    model_name="BAAI/bge-m3",
    model_kwargs={'device': 'cuda'}
)

QDRANT_URL = "https://3aef3ea0-03f1-4fce-ae5c-cc7d00ff4741.us-west-1-0.aws.cloud.qdrant.io/"
QDRANT_API_KEY = userdata.get('QDRANT_API_KEY')   # 🔄️colab에서 Qdrant api key get
# 🔄️COLLECTION_NAME = "welfare_chunks_bge_m3" => 새로운 collection으로 바꿨습니다.
COLLECTION_MINISTRY = "welfare_ministry_bge_m3"
COLLECTION_LOCAL = "local_government_welfare_bge_m3"

client = QdrantClient(url=QDRANT_URL, api_key=QDRANT_API_KEY)

# 중앙부처용 DB 연결
vector_db_ministry = QdrantVectorStore(
    client=client,
    collection_name=COLLECTION_MINISTRY,
    embedding=embeddings,
    content_payload_key="text"
)

# 지자체용 DB 연결
vector_db_local = QdrantVectorStore(
    client=client,
    collection_name=COLLECTION_LOCAL,
    embedding=embeddings,
    content_payload_key="text"
)
print("✅ 중앙부처 & 지자체 DB 모두 연결 완료!\n")

# =====================================================================
# [단계 5] URL 누락 방지 완벽 패치 및 질의응답 함수
# =====================================================================

# 🔄️ LLM 프롬프트 코드 직전까지(def부분부터 데이터 정제 파트까지) 간결하게 바꿨습니다. qdrant metadata 관련 최적
def ask_with_provenance(query, vector_db_ministry, vector_db_local, model, tokenizer):
    # 1. 두 곳에서 각각 검색 (각각 3개씩, 총 6개 후보)
    docs_ministry = vector_db_ministry.similarity_search(query, k=3)
    docs_local = vector_db_local.similarity_search(query, k=3)

    # 2. 결과 합치기
    all_docs = docs_ministry + docs_local

    if not all_docs:
        return "죄송합니다. 관련 정보를 찾지 못했습니다."

    # 3. 데이터 정제 (사용자 정의 필드 활용)
    final_context_list = []
    source_info = []

    for doc in all_docs:
        m = doc.metadata
        # welfare_chunks.jsonl의 실제 키 값에 맞춰 매핑
        title = m.get("제목") or m.get("policy_name", "제목 없음")
        content = doc.page_content # 상세 내용 (text 필드)
        url = m.get("신청 링크") or m.get("url", "정보 없음")
        region = m.get("필터링") or m.get("region", "전국")

        final_context_list.append(f"정책명: {title}\n지역: {region}\n내용: {content}")
        source_info.append(f"  - [{region}] {title}: {url}")

    final_context = "\n\n".join(final_context_list)

# model_service.py에 있던 함수 추가
def analyze_with_captum(
        self,
        user_input: str,
        history: Optional[List[Dict[str, str]]] = None,
    ) -> Dict[str, Any]:
        """
        사용자 원본 질문을 Captum으로 분석한다.
        반환 형식은 기존 chat.js와 맞춘다:
        { target_token_id, target_token, word_scores, delta }
        """
        try:
            enc = self.tokenizer(
                user_input,
                return_tensors="pt",
                add_special_tokens=False,
                truncation=True,
                max_length=128,
                return_offsets_mapping=True,
            )

            input_ids = enc["input_ids"].to(self.device)
            attention_mask = enc["attention_mask"].to(self.device)
            offset_mapping = enc["offset_mapping"][0].tolist()

            with torch.no_grad():
                logits = self.captum_forward_func(input_ids, attention_mask)
                target_token_id = int(torch.argmax(logits, dim=-1).item())

            pad_token_id = self.tokenizer.pad_token_id
            if pad_token_id is None:
                pad_token_id = self.tokenizer.eos_token_id

            baseline_ids = torch.full_like(input_ids, pad_token_id)

            self.model.zero_grad(set_to_none=True)

            attributions, delta = self.lig.attribute(
                inputs=input_ids,
                baselines=baseline_ids,
                additional_forward_args=(attention_mask,),
                target=target_token_id,
                return_convergence_delta=True,
                n_steps=16,
                internal_batch_size=1,
            )

            token_attributions = attributions.sum(dim=-1).squeeze(0)
            token_attributions = token_attributions.detach().float().cpu().tolist()

            word_matches = list(re.finditer(r"\S+", user_input))
            word_scores: List[Dict[str, Any]] = []

            for match in word_matches:
                w_start, w_end = match.span()
                word_text = match.group()

                # 문장부호만 있는 토큰은 하이라이트 대상에서 제외한다.
                if not re.search(r"[0-9A-Za-z가-힣]", word_text):
                    continue

                score_sum = 0.0
                for (t_start, t_end), t_score in zip(offset_mapping, token_attributions):
                    if t_start == t_end:
                        continue

                    overlap = not (t_end <= w_start or t_start >= w_end)
                    if overlap:
                        score_sum += abs(float(t_score))

                word_scores.append(
                    {
                        "word": word_text,
                        "start": w_start,
                        "end": w_end,
                        "score": round(score_sum, 6),
                    }
                )

            max_score = max((w["score"] for w in word_scores), default=0.0)
            if max_score > 0:
                for w in word_scores:
                    w["score"] = round(w["score"] / max_score, 6)

            target_token = self.tokenizer.decode(
                [target_token_id],
                skip_special_tokens=True,
                clean_up_tokenization_spaces=False,
            ).strip()

            delta_value = delta.detach().float().cpu()
            if delta_value.numel() == 1:
                delta_out: Any = float(delta_value.item())
            else:
                delta_out = delta_value.tolist()

            return {
                "target_token_id": target_token_id,
                "target_token": target_token,
                "word_scores": word_scores,
                "delta": delta_out,
            }

        finally:
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            gc.collect()

    # 4. LLM 프롬프트
    prompt = f"""<|im_start|>system
당신은 대한민국 복지 정책을 안내하는 친절하고 정확한 상담사입니다.
주어진 [문서] 내용만을 바탕으로 답변하고, 100% 한국어로만 작성하세요. 한자나 영어는 절대 금지합니다.

[답변 작성 규칙]
1. 정책의 핵심 내용(대상, 지원내용 등)을 사람이 말하듯 자연스럽게 설명하세요.
2. 상세 URL이나 링크는 사용자가 이미 화면에서 확인했으므로, 답변에 절대 링크나 URL을 생성하지 마세요. (가짜 링크 방지)
3. 만약 문서에 정보가 없다면 '죄송합니다. 관련 정보를 찾지 못했습니다.'라고만 답하세요.
4. 문서에 없는 구체적인 서비스 목록이나 혜택을 당신이 상상해서 1, 2, 3번으로 나열하지 마세요.
5. 한자(漢字), 일본어, 영어는 절대 사용하지 마세요. 100% 순수 한국어로만 작성하세요.
6. 묻는 말에만 3~4문장 이내로 짧고 명확하게 답변하세요.
<|im_end|>
<|im_start|>user
[문서]
{final_context}

질문: {query}<|im_end|>
<|im_start|>assistant
"""

    # 5. 답변 생성
    device = next(model.parameters()).device
    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    print("🤖 답변을 생성 중입니다...\n")
    outputs = model.generate(
        **inputs,
        max_new_tokens=512,
        temperature=0.1,
        repetition_penalty=1.1,
        pad_token_id=tokenizer.pad_token_id,
        eos_token_id=tokenizer.eos_token_id
    )

    answer = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True).strip()

    print(f"🤖 Qwen 3B AI 답변:\n{answer}\n" + "="*60)

    return answer

# 무한 루프 제거했습니다.
```

```python
from fastapi import FastAPI
from pydantic import BaseModel
import uvicorn
import asyncio
import nest_asyncio
import traceback

nest_asyncio.apply()
app = FastAPI()

class ChatRequest(BaseModel):
    query: str

# Query api
@app.post("/ask")
def ask_api(request: ChatRequest):
    try:
        print(f"질문 도착: {request.query}")
        answer = ask_with_provenance(
            request.query,
            vector_db_ministry,
            vector_db_local,
            model,
            tokenizer
        )
        return {"answer": answer}

    except Exception as e:
        error_trace = traceback.format_exc()
        print(f"에러 발생:\n{error_trace}")
        return {"answer": f"에러 원인:\n{error_trace}"}

# Analyzation API
@app.post("/analyze")
def analyze_api(request: ChatRequest):
    try:
        print(f"XAI 분석 요청 도착: {request.query}")

        # 기존 model_service.py에 있던 분석 함수를 여기서 실행
        result = analyze_with_captum(request.query)

        return result  # {"words": [...], "attributions": [...]} 형태 반환

    except Exception as e:
        error_trace = traceback.format_exc()
        print(f"XAI 분석 에러 발생:\n{error_trace}")
        return {"words": ["분석", "중", "에러가", "발생했습니다."], "attributions": [0.0, 0.0, 0.0, 0.0]}

if __name__ == "__main__":
    loop = asyncio.get_event_loop()

    # 1. 백그라운드의 기존 서버 태스크를 찾아 취소(Kill)
    for task in asyncio.all_tasks(loop):
        if task.get_name() == 'fastapi_server':
            task.cancel()
            print("기존 서버를 종료 및 메모리 비우기 완료")

    # 2. 약간 대기 후 새 코드로 서버 다시 시작
    config = uvicorn.Config(app, host="0.0.0.0", port=8000, log_level="info")
    server = uvicorn.Server(config)
    loop.create_task(server.serve(), name='fastapi_server')
    print("에러 추적기가 탑재된 새 서버가 시작")
    print("접속 주소: https://kr-welfare-xai.com/ask")
```
